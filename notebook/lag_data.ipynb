# 1. Import & Setup

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.impute import SimpleImputer
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
```

# 2. Daten einlesen

```python
df_raw = pd.read_csv(
    "../data/Number_of_fires_by_month.csv",
    sep=";",
    skiprows=1
)

df_raw.head()
```

# 3. Relevante Spalten behalten

```python
df = df_raw.drop(columns=["Data Qualifier"])
```

# 4. Wide → Long (DER entscheidende Schritt)

```python
df_long = df.melt(
    id_vars=["Jurisdiction", "Month"],
    var_name="Year",
    value_name="Number_fires"
)
```

# 5. Aufräumen & Typen fixen

```python
df_long["Year"] = df_long["Year"].astype(int)
df_long["Number_fires"] = pd.to_numeric(
    df_long["Number_fires"],
    errors="coerce"
)
```

# 6. Monate in Zahlen übersetzen

```python
month_map = {
    "January": 1, "February": 2, "March": 3,
    "April": 4, "May": 5, "June": 6,
    "July": 7, "August": 8, "September": 9,
    "October": 10, "November": 11, "December": 12
}

df_long["Month"] = df_long["Month"].map(month_map)
```

# 6. Nur British Columbia behalten

```python
df_bc = df_long[df_long["Jurisdiction"] == "British Columbia"]
df_bc = df_bc.sort_values(["Year", "Month"])
```

# 7. Das echte Lag-CSV

```python
df_bc[["Year", "Month", "Jurisdiction", "Number_fires"]].to_csv(
    "fires_history_bc.csv",
    index=False
)
```
