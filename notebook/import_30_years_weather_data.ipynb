### Wetter Daten laden (Von 1990 bis 2023)

### British Columbia (BC)

```python
#Cariboo Region (Williams Lake Area)
#→ 52.15, -122.15

import pandas as pd

pfad = "../data/import weather data/import data/British Columbia/open-meteo-52.55N122.18W978m-2.csv"

# -------------------------------------------------------
# 1) Koordinaten aus Zeile 2 (Index 1) einlesen
# -------------------------------------------------------
coords = pd.read_csv(pfad, nrows=2, header=None)
lat = coords.iloc[1, 0]
lon = coords.iloc[1, 1]

df = pd.read_csv(
    pfad,
    skiprows=3,   # Metadaten überspringen
    comment="#",  # ignoriert eventuelle Kommentarzeilen
)

df["latitude"] = lat
df["longitude"] = lon

# Falls du die Spalten vorne haben willst:
British_Columbia_1 = df[["latitude", "longitude"] + [col for col in df.columns if col not in ["latitude", "longitude"]]]

British_Columbia_1
```

```python
#Okanagan – feueranfälligste Region in BC
#→ 49.80, -119.60

pfad = "../data/import weather data/import data/British Columbia/open-meteo-49.81N119.69W339m-2.csv"

# -------------------------------------------------------
# 1) Koordinaten aus Zeile 2 (Index 1) einlesen
# -------------------------------------------------------
coords = pd.read_csv(pfad, nrows=2, header=None)
lat = coords.iloc[1, 0]
lon = coords.iloc[1, 1]

df = pd.read_csv(
    pfad,
    skiprows=3,   # Metadaten überspringen
    comment="#",  # ignoriert eventuelle Kommentarzeilen
)

df["latitude"] = lat
df["longitude"] = lon

# Falls du die Spalten vorne haben willst:
British_Columbia_2 = df[["latitude", "longitude"] + [col for col in df.columns if col not in ["latitude", "longitude"]]]

British_Columbia_2
```

```python
#Peace River Forest / Fort St. John Hinterland
#→ 56.30, -121.00


pfad = "../data/import weather data/import data/British Columbia/open-meteo-56.27N121.05W789m-2.csv"

# -------------------------------------------------------
# 1) Koordinaten aus Zeile 2 (Index 1) einlesen
# -------------------------------------------------------
coords = pd.read_csv(pfad, nrows=2, header=None)
lat = coords.iloc[1, 0]
lon = coords.iloc[1, 1]

df = pd.read_csv(
    pfad,
    skiprows=3,   # Metadaten überspringen
    comment="#",  # ignoriert eventuelle Kommentarzeilen
)

df["latitude"] = lat
df["longitude"] = lon

# Falls du die Spalten vorne haben willst:
British_Columbia_3 = df[["latitude", "longitude"] + [col for col in df.columns if col not in ["latitude", "longitude"]]]

British_Columbia_3
```

### Daten zusammenführen pro Provinz und Durchschnitt berechnen

```python

# in Datetime umwandeln
for df in [British_Columbia_1, British_Columbia_2, British_Columbia_3]:
    df['time'] = pd.to_datetime(df['time'])

#  drei DataFrames zusammenführen
bc_all = pd.concat(
    [British_Columbia_1, British_Columbia_2, British_Columbia_3],
    ignore_index=True
)

# Wetterspalten definieren
weather_columns = [
    'temperature_2m_mean (°C)',
    'temperature_2m_max (°C)',
    'relative_humidity_2m_mean (%)',
    'relative_humidity_2m_min (%)',
    'precipitation_sum (mm)',
    'precipitation_hours (h)',
    'wind_speed_10m_mean (km/h)',
    'wind_speed_10m_max (km/h)',
    'wind_direction_10m_dominant (°)',
    'et0_fao_evapotranspiration (mm)'
]

# Tagesdurchschnitt über alle 3 Koordinatenpunkte berechnen
British_Columbia_mean = (
    bc_all
    .groupby('time')[weather_columns]
    .mean()
    .reset_index()
)

British_Columbia_mean
```

### Durschnitt pro Monat


```python

British_Columbia_mean['time'] = pd.to_datetime(British_Columbia_mean['time'])

# Monats-Durchschnitt für alle Variablen
British_Columbia_monthly_mean = (
    British_Columbia_mean
    .set_index('time')
    .resample('MS')
    .mean()
    .reset_index()
)

British_Columbia_monthly_mean
```

```python
import os
#Datei speichern
output_path = "../data/cleaned_British_Columbia_monthly_mean.csv"


# Sicherstellen, dass das Verzeichnis existiert
os.makedirs(os.path.dirname(output_path), exist_ok=True)

British_Columbia_monthly_mean.to_csv(output_path, index=False)
```
